```julia
using LowLevelFEM, LinearAlgebra
```

```julia
#openGeometry("mpc-1.geo")
structured_rect_mesh(lx=10)
Premote = gmsh.model.occ.addPoint(10, 0.5, 0)
gmsh.model.occ.synchronize()
gmsh.model.addPhysicalGroup(0, [Premote], -1, "remote")
gmsh.model.mesh.clear()
gmsh.model.mesh.generate(3)

#openPreProcessor()
```

```julia
mat1 = Material("body")
mat2 = Material("remote")

U = Field([mat1, mat2], type=:VectorField, dim=2, fieldName=:u, rhsName=:f)
```

```julia
K = ∫(SymGrad(U) ⋅ D(:PlaneStress, mat1) ⋅ SymGrad(U), Ω="body")
K[:,:]
```

```julia
M = ∫(U ⋅ mat1.ρ ⋅ U, Ω="body")
M[:,:]
```

```julia
mpc = MultiPointConstraint(master="remote", slave="right", ux=false)

bc = BoundaryCondition("left", ux=0, uy=0)
bc2 = BoundaryCondition("remote", ux=0)
```

```julia
f = ∫(U ⋅ [0, 1], Γ="remote")
```

```julia
smallestEigenValue(K, M, support=[bc2], mpc=[mpc])
```

```julia
smallestEigenValue(K, M, support=[bc,bc2], mpc=[mpc])
```

```julia
largestEigenValue(K, M, support=[bc2], mpc=[mpc])
```

```julia
largestEigenValue(K, M, support=[bc,bc2], mpc=[mpc])
```

```julia
UU = solveEigenModes(K, M, support=[bc,bc2], mpc=[mpc])
```

```julia
showModalResults(UU)
```

```julia
λmax = largestEigenValue(K, M, support=[bc,bc2], mpc=[mpc])
dt = 2 / √λmax
```

```julia
u = solveField(K, f, support=[bc,bc2], mpc=[mpc])
```

```julia
showDoFResults(u, name="u", visible=true, factor=100)
```

```julia
u2, v2 = HHT(K, M, f, 0u, 0u, 100, dt, support=[bc2], mpc=[mpc])
```

```julia
showDoFResults(u2, factor=10000)
```
