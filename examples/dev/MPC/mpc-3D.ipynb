```julia
using LowLevelFEM
```

```julia
openGeometry("mpc-2.geo")

#openPreProcessor()
```

```julia
mat1 = Material("body")
mat2 = Material("remote")

U = Field([mat1, mat2], type=:VectorField, dim=3, fieldName=:u, rhsName=:f)
Φ = Field([mat1, mat2], type=:VectorField, dim=3, fieldName=:φ, rhsName=:m)
```

```julia
E = mat1.E
ν = mat1.ν

D = E / (1+ν) / (1-2ν) * [1-ν ν ν 0 0 0; ν 1-ν ν 0 0 0; ν ν 1-ν 0 0 0; 0 0 0 (1-2ν)/2 0 0; 0 0 0 0 (1-2ν)/2 0; 0 0 0 0 0 (1-2ν)/2]
```

```julia
Ku = ∫(SymGrad(U) ⋅ D ⋅ SymGrad(U), Ω="body")
Ku[:,:]
```

```julia
mpc_u = MPC(master="remote", slave="right", field=U)

mpc_φ = MPC(master="remote", slave="right", field=Φ)
```

```julia
R = rigidRotationMap(mpc_u, mpc_φ)
R[:,:]
```

```julia
Kuφ = Ku * R
Kuφ[:,:]
```

```julia
Kφ = R' * Ku * R
Kφ[:,:]
```

```julia
K = SystemMatrix([Ku Kuφ; Kuφ' Kφ])
K[:,:]
```

```julia
bc = BoundaryCondition("left", field=U, ux=0, uy=0, uz=0)

fu = ∫(U ⋅ [0, 0, 0], Γ="remote")
fφ = ∫(Φ ⋅ [0, 100000, 0], Γ="remote")

F = SystemVector([fu, fφ])
```

```julia
u0, φ = solveField(K, F, support=[bc], mpc=[mpc_u, mpc_φ])
```

```julia
u = u0 + R * φ
```

```julia
showDoFResults(u, name="u", visible=true)
```

```julia
probe(φ, "remote")
```

```julia
bc2 = BoundaryCondition("remote", field=Φ, φz=0.0082760692)

fu = ∫(U ⋅ [0, 0, 0], Γ="remote")
fφ = ∫(Φ ⋅ [0, 0, 0], Γ="remote")

F = SystemVector([fu, fφ])
```

```julia
u0, φ = solveField(K, F, support=[bc, bc2], mpc=[mpc_u, mpc_φ])
```

```julia
u = u0 + R * φ
```

```julia
showDoFResults(u, name="u2")
```

```julia
probe(φ, "remote")
```
