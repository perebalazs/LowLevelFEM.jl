```julia
using LowLevelFEM
```

```julia
openGeometry("2D-Hertz.geo")
#openPreProcessor()
```

```julia
matU = Material("upper", E=2e5, ν=0.3)
matL = Material("lower", E=2e5, ν=0.3)

#renumberNodes!([matU, matL])

U = Field([matU, matL], type=:VectorField, dim=2, fieldName=:u);
```

```julia
bcU = BoundaryCondition("top", ux=0, uy=-0.0)
bcL1 = BoundaryCondition("bottom", uy=0)
bcL2 = BoundaryCondition("left", ux=0)
bcL3 = BoundaryCondition("right", ux=0)

bc = [bcU, bcL1, bcL2, bcL3];
```

```julia
K = ∫(ε(U) ⋅ D(:PlaneStrain, matL) ⋅ ε(U), Ω="lower") + ∫(ε(U) ⋅ D(:PlaneStrain, matU) ⋅ ε(U), Ω="upper")
K[:,:]
```

```julia
f = ∫(U ⋅ [0, 0], Γ="top");
```

```julia
u0 = solveField(K, f, support=bc);
showDoFResults(u0, name="u0", factor=1)
```

```julia
C = contact(U, master="master", slave="slave", normal_sign=-1)
```

```julia
G = ContactGap(C)
```

```julia
gap = G(u0)
gap = mapScalarField((x)->min(x, 0.0), gap)
plotOnBeam("slave", nodesToElements(gap), name="gap")
```

```julia
cn = 1e6
```

```julia
Kc = ∫(G ⋅ cn ⋅ G)
Kc[:,:]
```

```julia
r = nodePositionVector(U)
r = projectTo2D(r)
r.a
```

```julia
r = vectorField(U, "upper", [(x,y,z)->x, (x,y,z)->y, (x,y,z)->z])
r.a
```

```julia
#r = nodePositionVector(U)
r = vectorField(U, "upper", [(x,y,z)->x, (x,y,z)->y, (x,y,z)->z]) + vectorField(U, "lower", [(x,y,z)->x, (x,y,z)->y, (x,y,z)->z])
fc = Kc * r
```

```julia
u1 = solveField(K + Kc, 0f - fc, support=bc);
```

```julia
showDoFResults(u1, name="u1", factor=1)
```
