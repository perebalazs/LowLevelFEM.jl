```julia
using LowLevelFEM
using Arpack
```

```julia
l = 100 # mm
n = 80 # pcs
v_0 = 1000 # mm/s
E = 90 # MPa
ρ = 7.85e-9 # t/mm^3
A = 1 # mm^2
order = 9;
```

```julia
line_mesh(lx=l, n=n, order=order)
```

```julia
mat = Material("body", E=E, ρ=ρ)
U = Field([mat], type=:ScalarField, dim=1, fieldName=:u);
```

```julia
K = ∫(Grad(U) ⋅ (E * A) ⋅ Grad(U), gauss=:full)

M = (ρ * A) * ∫(U ⋅ U, gauss=:full)
M = consistentToLumped(M)

f = ∫(U ⋅ 0);
```

```julia
bc = BoundaryCondition("right", u=0)

u0 = scalarField(U, "body", 0)
v0 = scalarField(U, "body", v_0);
```

```julia
dt = l / √(E / ρ) / n / order
```

```julia
t_end = 2l / sqrt(E/ρ)
steps = ceil(Int, t_end / dt) + 1
```

```julia
u, v = HHT(K, M, f, u0, v0, steps, dt, support=[bc], α=0, γ=0.5, β=0.25);
```

```julia
u1 = showDoFResults(u, name="u fem")
v1 = showDoFResults(v, name="v fem")
```

```julia
plotOnPath("body", u1, name="u plot", points=2n*order)
plotOnPath("body", v1, name="v plot", points=2n*order)
```

```julia
p = E * ∂x(u)

p1 = showElementResults(p, name="p")
```

```julia
plotOnPath("body", p1, points=2n*order, name="p plot")
```
