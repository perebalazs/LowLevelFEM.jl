```julia
using LowLevelFEM
```

```julia
openGeometry("multi-part.geo")
#openPreProcessor()
```

```julia
mat1 = Material("part1")
mat2 = Material("part2")
P1 = Field([mat1], type=:VectorField, dim=3)
P2 = Field([mat2], type=:VectorField, dim=3)
```

```julia
v1 = VectorField(P1, "part1", [1,1,1])
v2 = VectorField(P2, "part2", [2,2,2])
```

```julia
@show P1.non
@show P2.non
```

```julia
showElementResults(v1, name="v1")
showElementResults(v2, name="v2")
```

```julia
v3 = v1 + v2
v3.model
```

```julia
v4 = v2 + v1
v4.model
```

```julia
showElementResults(v3, name="v3")
showElementResults(v4, name="v4")
```

```julia
v5 = elementsToNodes(v1)
v6 = elementsToNodes(v2)
```

```julia
showDoFResults(v5, name="v5")
showDoFResults(v6, name="v6")
```

```julia
d1 = allDoFs(P1)
```

```julia
d2 = allDoFs(P2)
```

```julia
v5 *= 0
v5.a[d1] .= 1

v6 *= 0
v6.a[d2] .= 2;
```

```julia
showDoFResults(v5, name="v5")
showDoFResults(v6, name="v6")
```
