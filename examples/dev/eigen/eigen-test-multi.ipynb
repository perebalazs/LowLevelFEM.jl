```julia
using LowLevelFEM
using LinearAlgebra
using SparseArrays
using Test

spectral_path = joinpath(
    dirname(pathof(LowLevelFEM)),
    "eigen.jl"
)

@assert isfile(spectral_path) "Copy spectral.jl into the LowLevelFEM src directory first: $spectral_path"

Base.include(
    LowLevelFEM,
    spectral_path
)

using LowLevelFEM: spectralTransformation
```

```julia
structured_rect_mesh(lx=10, n=2)

mat = Material("body")
U = Field([mat], type=:VectorField, dim=2, fieldName=:u);
V = Field([mat], type=:VectorField, dim=2, fieldName=:v);
```

```julia
ndofs(U)
```

```julia
Ku = ∫(ε(U) ⋅ D(:PlaneStress, mat) ⋅ ε(U))
Kuv = 0 * ∫(ε(U) ⋅ D(:PlaneStress, mat) ⋅ ε(V))
Kv = ∫(ε(V) ⋅ D(:PlaneStress, mat) ⋅ ε(V))

Mu = ∫(U ⋅ mat.ρ ⋅ U)
Muv = 0 * ∫(U ⋅ mat.ρ ⋅ V)
Mv = ∫(V ⋅ mat.ρ ⋅ V)

K = SystemMatrix([Ku Kuv; Kuv' Kv])
M = SystemMatrix([Mu Muv; Muv' Mv])
K[:,:]
```

```julia
bc1 = BoundaryCondition("left", ux=0, uy=0, field=U)
bc2 = BoundaryCondition("left", vx=0, vy=0, field=V)
```

```julia
λ, Φ = solveEigen(Symmetric(K), Symmetric(M), solver=:direct, output=:raw)
λ
```

```julia
λ, Φ = solveEigen(Symmetric(K), Symmetric(M), solver=:arpack, output=:raw)
λ
```

```julia
λ, Φ = solveEigen(Symmetric(K), Symmetric(M), support=[bc1,bc2], solver=:direct, output=:raw)
λ
```

```julia
λ, Φ = solveEigen(Symmetric(K), Symmetric(M), shift=0, nev=4, solver=:arpack, output=:raw)
λ
```

```julia
λ, Φ = solveEigen(Symmetric(K), Symmetric(M), support=[bc1,bc2], output=:raw)
λ
```

```julia
λ, ϕ = solveEigen(
    Symmetric(K),
    Symmetric(M);
    support=[bc1,bc2],
    n=6,
    shift=0.0,
    solver=:arpack,
    solveroptions=(ncv=40, maxiter=10000), output=:raw
)

λ
```

```julia
λ, ϕ = solveEigen(
    Symmetric(K),
    Symmetric(M);
    support=[bc1,bc2],
    n=6,
    sigma=0.0,
    which=:LM,
    solver=:arpack,
    solveroptions=(
        explicittransform=:none,
        ncv=40,
        maxiter=10000
    ), output=:raw
)

@show λ
@show sqrt.(real.(λ)) ./ (2π)
```

```julia
λ, ϕ = solveEigen(Symmetric(K), Symmetric(M), sigma=1, which=:LM, output=:raw)
λ
```

```julia
λ, ϕ = solveEigen(K, M, support=[bc1,bc2], output=:raw)
λ
```

```julia
P = LowLevelFEM.prepare_eigen_system(
    K, M;
    support=[bc1,bc2]
)

Kr, Mr = P.A, P.B

@show size(Kr)
@show nnz(Kr), nnz(Mr)

@show norm(Kr - Kr')
@show norm(Mr - Mr')

@show extrema(diag(Kr))
@show extrema(diag(Mr))

@show isposdef(Symmetric(Kr))
@show isposdef(Symmetric(Mr))
```

```julia
per1 = MultiPointConstraint(master="left", slave="right", field=U)
per2 = MultiPointConstraint(master="left", slave="right", field=V)
```

```julia
e1 = VectorField(U, "left", [-1, 0, 0])
cs = CoordinateSystem(e1)
```

```julia
λ, Φ = solveEigen(Symmetric(K), Symmetric(M), coordSys=[cs], mpc=[per1,per2], solver=:direct, output=:raw)
λ
```

```julia
λ, Φ = solveEigen(Symmetric(K), Symmetric(M), coordSys=[cs], mpc=[per1,per2], output=:raw)
λ
```

```julia
λ, Φ = solveEigen(Symmetric(K), Symmetric(M), coordSys=[cs], mpc=[per1,per2], solveroptions=(ncv=40, maxiter=10000), output=:raw)
λ
```
