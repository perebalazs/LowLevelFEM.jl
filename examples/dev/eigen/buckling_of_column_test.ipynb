```julia
mat = Material("body")
U = Field([mat], type=:VectorField, dim=3, fieldName=:u);
```

```julia
@time K = ∫(SymGrad(U) ⋅ D(:Solid, mat) ⋅ SymGrad(U))
K[:,:]
```

```julia
f = ∫(U ⋅ [-1, 0, 0], Γ="right");
```

```julia
bc = BoundaryCondition("left", ux=0, uy=0, uz=0);
```

```julia
u = solveField(Symmetric(K), f, support=[bc])
showDoFResults(u, name="u");
```

```julia
ε = (u ∘ ∇ + ∇ ∘ u) / 2
```

```julia
# ε2 = (∇(u)' + ∇(u)) / 2
# ∫(U, "body", (ε - ε2) ⋅ (ε - ε2))
```

```julia
E = mat.E
ν = mat.ν
I = unitTensor(ε)

σ = E / (1+ν) * (ε + ν / (1-2ν) * trace(ε) * I)
```

```julia
σ = TensorField(U, "body", [-1 0 0; 0 0 0; 0 0 0]);
```

```julia
DG = Matrix{Any}(undef, 9, 9)

fill!(DG, 0.0)

for c in 1:3, i in 1:3, j in 1:3
    DG[(c-1)*3+i, (c-1)*3+j] = σ[i,j]
end

@time Kg = ∫(Grad(U) ⋅ DG ⋅ Grad(U))
Kg[:,:]
```

```julia
#=
DG = Matrix{Any}(undef, 9, 9)

fill!(DG, 0.0)

DG[1,1] = -1.0
DG[4,4] = -1.0
DG[7,7] = -1.0

@time Kg = ∫(Grad(U) ⋅ DG ⋅ Grad(U))
Kg[:,:]
=#
```

```julia
u1 = solveEigen(Kg, K, support=[bc])
u1.t .= -1 ./ u1.t
```

```julia
dom = LowLevelFEM.DomainSpec(:Ω, "body")

# Warm-up
Cprep = LowLevelFEM._prepare_coefficient(DG, dom)

GC.gc()
@time Cprep = LowLevelFEM._prepare_coefficient(DG, dom)
```

```julia
using Profile

Profile.Allocs.clear()

Profile.Allocs.@profile sample_rate=0.0001 begin
    Kg = ∫(Grad(U) ⋅ DG ⋅ Grad(U); threads=1)
end

prof = Profile.Allocs.fetch()

@show length(prof.allocs)
```

```julia
counts = Dict{Any,Int}()
bytes  = Dict{Any,Int}()

for a in prof.allocs
    counts[a.type] = get(counts, a.type, 0) + 1
    bytes[a.type]  = get(bytes, a.type, 0) + a.size
end

sort(
    collect(counts);
    by=last,
    rev=true
)[1:min(15, length(counts))]
```

```julia
subarrays = filter(
    a -> occursin("SubArray", string(a.type)),
    prof.allocs
)

@show length(subarrays)

for fr in first(subarrays).stacktrace[1:12]
    println(fr)
end
```

```julia
using Statistics

expr = Grad(U) ⋅ DG ⋅ Grad(U)

function benchmark_assembly(expr; threads=:auto, n=5)
    ∫(expr; threads=threads)  # Warm-up

    times = Float64[]

    for _ in 1:n
        GC.gc()
        t = @elapsed ∫(expr; threads=threads)
        push!(times, t)
    end

    @show times
    @show median(times)
end

benchmark_assembly(expr; threads=:auto)
benchmark_assembly(expr; threads=1)
```
