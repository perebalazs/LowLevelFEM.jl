```julia
using LowLevelFEM, LinearAlgebra
```

```julia
structured_rect_mesh(lx=1, ly=1, n=100, order=2)

mat = Material("body")
V = Problem([mat], type=:ScalarField, dim=2, field=:v, bandwidth=:Metis)
S = Problem([mat], type=:ScalarField, dim=2, field=:s);
```

```julia
E = mat.E
ρ = mat.ρ
A = 4
v0 = 1
```

```julia
I = [
    1.0 0.0
    0.0 1.0
]

C = [
    0.0 -1.0
    -1.0 0.0
]

Kvv = ∫(Grad(V) ⋅ I ⋅ Grad(V))
Kvs = ∫(Grad(V) ⋅ C ⋅ Grad(S))

Ksv = ∫(Grad(S) ⋅ C ⋅ Grad(V))
Kss = ∫(Grad(S) ⋅ I ⋅ Grad(S))
```

```julia
K = SystemMatrix([
    Kvv Kvs
    Ksv Kss
])
```

```julia
v0 = ScalarField(V, "body", (x, y, z)->exp(-100*(x-0.5)^2))
```

```julia
bc_v = BoundaryCondition("bottom", problem=V, v=v0)
bc_s = BoundaryCondition("bottom", problem=S, s=0.0)
```

```julia
f_v = ∫(V ⋅ 0.0)
f_s = ∫(S ⋅ 0.0)

f = SystemVector([f_v, f_s])
```

```julia
v, s = solveField(
    K,
    f;
    support=[bc_v, bc_s]
)
```

```julia
showDoFResults(v, name="v(x,t)", visible=true)
showDoFResults(s, name="s(x,t)")
```
