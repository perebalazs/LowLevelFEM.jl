```julia
using LowLevelFEM, LinearAlgebra
```

```julia
L = 1
```

```julia
structured_rect_mesh(lx=L, ly=2, n=50, order=1)

mat = Material("body")
renumberNodes!(mat, method=:RCMK)
V = Field([mat], type=:ScalarField, dim=2, field=:v)
S = Field([mat], type=:ScalarField, dim=2, field=:s);
```

```julia
E = mat.E
ρ = mat.ρ
A = 4
v_0 = 1
```

```julia
I = [
    1.0 0.0
    0.0 1.0
]

C = [
    0.0 -1.0
    -1.0 0.0
]

Kvv = ∫(Grad(V) ⋅ I ⋅ Grad(V))
Kvs = ∫(Grad(V) ⋅ C ⋅ Grad(S))

Ksv = ∫(Grad(S) ⋅ C ⋅ Grad(V))
Kss = ∫(Grad(S) ⋅ I ⋅ Grad(S))
```

```julia
K = SystemMatrix([
    Kvv Kvs
    Ksv Kss
])
K[:, :]
```

```julia
v0 = ScalarField(V, "bottom", (x, y, z)->x == L ? 0 : v_0)
```

```julia
bc_v1 = BoundaryCondition("bottom", problem=V, v=v0)
bc_v2 = BoundaryCondition("right", problem=V, v=0)
bc_s1 = BoundaryCondition("bottom", problem=S, s=0.0)
bc_s2 = BoundaryCondition("left", problem=S, s=0.0)
```

```julia
f_v = ∫(V ⋅ 0.0)
f_s = ∫(S ⋅ 0.0)

f = SystemVector([f_v, f_s])
```

```julia
v, s = solveField(
    K,
    f;
    support=[bc_v1, bc_v2, bc_s1, bc_s2]
)
```

```julia
showDoFResults(v, name="v(x,t)", visible=true)
showDoFResults(s, name="s(x,t)")
```
