```julia
using LowLevelFEM, LinearAlgebra, SparseArrays
```

```julia
openGeometry("beam-1.geo")
#openPreProcessor()

mat = Material("beam")
Pu = Problem([mat], type=:VectorField, dim=2, field=:u, rhs_field=:f)
Pφ = Problem([mat], type=:ScalarField, dim=2, field=:φ, rhs_field=:m, reducedOrder=true)
```

```julia
a = 80
v = 2
E = mat.E
G = mat.μ
Iz = a^4 / 12 - (a - 2v)^4 / 12
κ = 2 / 3
A = a * a - (a - 2v)^2
```

```julia
t = tangentVector(Pu, "beam")
tx, ty, tz = t[1], t[2], t[3]

nx = -ty
ny = tx
nz = 0
```

```julia
showElementResults(t)
showElementResults(VectorField([nx, ny, 0ny]))
```

```julia
a = [tx * tx tx * ty ty * tx ty * ty]   # axial projector
s = [nx * tx nx * ty ny * tx ny * ty]   # shear projector
```

```julia
Cuu = E * A * (a' * a) + κ * G * A * (s' * s)

Cuφ = -κ * G * A * s'   # 4×1
Cφu = -κ * G * A * s    # 1×4
```

```julia
Kuu = ∫(Grad(Pu) ⋅ Cuu ⋅ Grad(Pu); Γ="beam")

Kuφ = ∫(Grad(Pu) ⋅ Cuφ ⋅ Id(Pφ); Γ="beam")
Kφu = ∫(Id(Pφ) ⋅ Cφu ⋅ Grad(Pu); Γ="beam")

Kφφ = ∫(Id(Pφ) ⋅ (κ * G * A) ⋅ Id(Pφ); Γ="beam") +
      ∫(AxialGrad(Pφ) ⋅ (E * Iz) ⋅ AxialGrad(Pφ); Γ="beam")
```

```julia
K = SystemMatrix([Kuu Kuφ; Kφu Kφφ])
```

```julia
supp_u = BoundaryCondition("A", problem=Pu, ux=0, uy=0)
supp_φ = BoundaryCondition("A", problem=Pφ, φ=0)
```

```julia
fu = ∫(Pu ⋅ [1, 0], Γ="C")
fφ = ∫(Pφ ⋅ 0, Γ="B")
fu2 = ∫(Pu ⋅ [0.0, -0.0], Γ="beam")
```

```julia
F = SystemVector([fu + fu2, fφ])
```

```julia


Tφ, Rφ = reductionMatrices(Pφ)

Tu = spdiagm(0 => ones(Pu.non*Pu.pdim))
TT = blockdiag(Tu, Tφ)

Kr = TT' * K.A * TT

eigvals(Matrix(Kr))
```

```julia
ndofs(P::Problem) = P.non * P.pdim
```

```julia
Tφ, Rφ = reductionMatrices(Pφ)

Tu = spdiagm(0 => ones(ndofs(Pu)))
Ru = Tu

T = blockdiag(Tu, Tφ)
R = blockdiag(Ru, Rφ)

fixed = constrainedDoFs(K, [supp_u, supp_φ])
free = freeDoFs(K, [supp_u, supp_φ])
xD = zeros(size(R, 2))

free_r, fixed_r, xD_r =
    reduced_bc_data(R, fixed, xD[:, 1])

Kr = T' * K.A * T

λ = eigvals(Symmetric(Matrix(Kr[free_r, free_r])))
λ[1:10]
```

```julia
λ = eigvals(Symmetric(Matrix(Kr[free_r, free_r])))

λmax = maximum(abs, λ)
tol = 1e-12 * λmax

count(abs.(λ) .< tol)
```

```julia
#u, v, φ = solveField(K, F, support=[supp_u, supp_v, supp_φ, supp_u0, supp_v0, supp_φ0])
u, φ = solveField(K, F, support=[supp_u, supp_φ])
```

```julia
showDoFResults(fu, name="force")
u0 = showDoFResults(u, name="u", factor=1, visible=true)
φ0 = showDoFResults(φ, name="φ")
```

```julia
n = VectorField([nx, ny, 0nx])
```

```julia
g = grad(expandTo3D(u))

ε = t ⋅ (g * t)
γ = n ⋅ (g * t) - nodesToElements(φ)

κb = grad(φ) ⋅ t

N = E * A * ε
T = κ * G * A * γ
Mh = E * Iz * κb
```

```julia
N0 = showElementResults(N, name="N")
T0 = showElementResults(T, name="T")
Mh0 = showElementResults(Mh, name="Mh")
```

```julia
plotOnPath("beam", N0)
plotOnPath("beam", T0)
plotOnPath("beam", Mh0)
```
