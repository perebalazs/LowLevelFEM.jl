```julia
using LowLevelFEM
```

```julia
openGeometry("periodic-2D.geo")
#openPreProcessor()
```

```julia
mat = Material("body")

U = Field([mat], type=:VectorField, dim=2, fieldName=:u);
```

```julia
K = ∫(ε(U)' ⋅ D(:PlaneStress, mat) ⋅ ε(U));
```

```julia
f = ∫(U ⋅ [0,0]);
```

```julia
bc = BoundaryCondition("left", ux=(x,y,z)->0.01*y, uy=0)

periodic = MPC(master="right", slave="left", uy=true)
```

```julia
u = solveField(K, f, support=[bc], mpc=[periodic])
```

```julia
showDoFResults(u, name="u", visible=true, factor=20)
```
